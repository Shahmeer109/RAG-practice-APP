### Data Ingestion

```python
from langchain_community.document_loaders import PyMuPDFLoader, TextLoader, JSONLoader, CSVLoader, DirectoryLoader
```

```python
pdf_loader = DirectoryLoader(
    "../data/pdf",
    glob="**/*.pdf",
    loader_cls=PyMuPDFLoader
)

csv_loader = DirectoryLoader(
    "../data/csv",
    glob="**/*.csv",
    loader_cls=CSVLoader
)

json_loader = DirectoryLoader(
    "../data/json",
    glob="**/*.json",
    loader_cls=JSONLoader,
    loader_kwargs={
        'jq_schema':".",
        'text_content':False
    }
)

text_loader = DirectoryLoader(
    "../data/text",
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding": "utf-8"}
)

documents = (
    pdf_loader.load()
    + csv_loader.load()
    + json_loader.load()
    + text_loader.load()
)
```

```python
for doc in documents:
    print("Content", doc.page_content)
    print("Metadata", doc.metadata)
    print("="*80)
```
